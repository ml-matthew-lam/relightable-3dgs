## Configuration

`DATA_DIR` points directly at the Drive-mounted dataset -- unlike training, this script only reads one small `transforms.json` file from it (for camera FOV), so there's no need to copy the full dataset to local disk first. If you'd rather not depend on the dataset at all, pass `--fov_deg <value>` in the render cell below instead and skip `DATA_DIR`.

```python
CHECKPOINT_DIR = "/content/drive/MyDrive/Personal Projects/relightable-3dgs/checkpoints/lambertian_week3"
DATA_DIR = "/content/drive/MyDrive/Personal Projects/relightable-3dgs/checkered_suzanne"
OUT_PATH = "/content/drive/MyDrive/Personal Projects/relightable-3dgs/videos/relight_tour.webp"

import os
os.makedirs(os.path.dirname(OUT_PATH), exist_ok=True)
```

## Rendering the tour video

Uses the latest checkpoint in `CHECKPOINT_DIR` by default (pass `--ckpt <path>` for a specific one). Adjust `--num_camera_positions` / `--camera_radius` / `--transition_frames` / `--dwell_frames` to taste -- this default set of args produces a ~6-position, ~2-second-per-transition, ~4-second-per-light-sweep tour.

```python
!python render_video.py \
  --ckpt_dir "{CHECKPOINT_DIR}" \
  --data_dir "{DATA_DIR}" \
  --num_camera_positions 6 \
  --camera_radius 4.5 \
  --camera_elevation_deg 20 \
  --light_radius 4.5 \
  --light_cycles 2 \
  --transition_frames 60 \
  --dwell_frames 90 \
  --fps 24 \
  --out "{OUT_PATH}"
```
